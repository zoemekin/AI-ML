1.3 ML-relevant warnings

Very important section.

You should discuss:

IDs are not features
playlist labels may be noisy proxies for musical style
popularity is not purely an audio phenomenon because marketing, artist fame, and platform effects matter
duplicates / repeated tracks across playlists may cause leakage or inflated performance
songs in playlists are curated, so the dataset is not a random sample of all music

This is exactly the kind of thinking that professors love.

# Block A: Genre

Main question

Can we predict playlist genre or subgenre from audio features alone?

This is stronger than a plain popularity regression because:

it is musically meaningful
it tests whether Spotify’s curated labels align with measurable sound patterns
it allows several model families
it connects directly to lectures on classification
Why this block matters

It asks whether genre is a real acoustic structure or mostly a human label.

Targets

You can choose:

primary: playlist_genre
secondary / harder: playlist_subgenre

Genre is easier and cleaner. Subgenre is more ambitious.

Models to compare

Use a progression of model families:

Zero-rule baseline
Logistic Regression
Decision Tree
Random Forest
SVM
maybe KNN as a distance-based contrast

That is already a strong lineup.

Why these models
Logistic Regression: interpretable linear baseline
Decision Tree: nonlinear rule-based model
Random Forest: robust ensemble, captures interactions
SVM: strong classifier in structured feature spaces
KNN: local similarity perspective

This is exactly the kind of justification your course wants.

Metrics

Since genre classes may be imbalanced:

accuracy
macro F1
weighted F1
confusion matrix

Macro F1 is especially important to show sensitivity to minority classes.

Interpretation goals
Which genres are easiest to distinguish?
Which genres overlap?
Does confusion happen between musically close categories?
Which features matter most for genre separation?
Is genre mostly linearly separable, or do nonlinear models win clearly?
Strong conclusion you may reach

If Random Forest / SVM strongly outperform Logistic Regression, that suggests genre boundaries in audio space are nonlinear or interaction-heavy.

# Block B: Popularity

Main question

To what extent can a song’s popularity be inferred from its audio profile?

The key here is not “predict popularity perfectly.”
The key is to show:

ambition
realism
limitations
thoughtful evaluation
Why this block matters

Popularity is tempting, but difficult and partially external to the song itself. That makes it intellectually interesting.

Important conceptual point

You should explicitly say:

popularity is influenced by non-audio factors not present in the dataset
therefore even a good model may have limited predictive power
this is not failure, it is evidence about the limits of the feature space

This is a very mature point.

Two ways to frame it
Option 1: Regression

Predict track_popularity as continuous target.

Models:

zero-rule regressor
Linear Regression / Ridge / Lasso
Decision Tree Regressor
Random Forest Regressor
Gradient Boosting Regressor if you want one beyond the lectures

This is solid.

Option 2: Popularity-tier classification

Convert popularity into bins:

low
medium
high

This can actually be stronger interpretively because:

easier to discuss
clearer metrics
aligns with classification methods
more robust than exact-point regression

Honestly, for a stylish notebook, I would consider doing both lightly, but making one primary.

My recommendation

Make tiered popularity classification the main supervised popularity task, and keep a small regression subsection as a supplementary check.

Why?
Because exact popularity is noisy, but popularity level is more interpretable.

Metrics

If classification:

macro F1
precision / recall per tier
confusion matrix

If regression:

MAE
RMSE
R²
Interpretation goals
Which audio properties correlate with popularity?
Are the most popular songs more danceable, louder, faster, more positive?
Do linear models already capture most of the signal, or do nonlinear models uncover hidden interactions?
Are there multiple distinct popularity profiles?
Very strong point

If results are only moderate, that is not a weak result. It supports the argument that popularity depends on much more than audio content.

# Block C: Audio features

Main question

What hidden musical groups emerge from the audio features, independent of Spotify’s labels?

This is probably the section that will make the notebook stand out.

Why this block matters

It goes beyond “predict the label.”
It asks:

what structure exists naturally in the data?
do curated genres align with actual clusters?
are there cross-genre musical archetypes?

That is exactly the kind of beyond-class ambition you want.

Methods
1. PCA

Use PCA first for:

dimensionality reduction
variance structure
visualization basis
maybe loadings to understand main audio axes

Interpret PCs like:

PC1 = energetic/loud vs acoustic/calm
PC2 = positive/danceable vs darker/speech-heavy
depending on results
2. K-Means clustering

Cluster songs in standardized audio space.

Try several k values and justify choice with:

silhouette score
elbow intuition
interpretability
3. Optional stronger extension

Use one more clustering approach for contrast:

Gaussian Mixture Model
or
DBSCAN if you want density-based argument

If you do this, the notebook becomes much richer. Even if only briefly.

What to analyze
cluster profiles by mean audio features
cluster distributions across genres
cluster distributions across popularity tiers
whether one genre spreads across many clusters
whether one cluster contains multiple genres
This is the gold

This lets you say things like:

“Genres are not acoustically pure categories”
“Songs from different genres share common sonic archetypes”
“Playlist labels reflect curation logic, but clustering reveals latent sound-based structure”
